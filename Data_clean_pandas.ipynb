```python
#I had to download pandas and matplotlib in order to import them
#Moved the given data into the same folder on jupyter notebook to be able to access them just by name
q_1 = flights_data[(flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')].shape[0]
print(f"q_1: Number of flights from JFK to SLC: {q_1}")
```

```python
#first need to find flights landing in slc
slc_flights = flights_data[flights_data['dest'] == 'SLC']
#find unique carriers for airlines
q_2 = slc_flights['carrier'].nunique()
print(f"q_2: {q_2} airlines fly to SLC.")
```

```python
#filter flights to rdu
rdu_flights = flights_data[flights_data['dest'] == 'RDU']
#calculate average arrival delay
q_3 = rdu_flights['arr_delay'].mean()
print(f"q_3: Average arrival delay for flights to RDU is {q_3:.2} minutes")
```

```python
#filter flights to sea
sea_flights = flights_data[flights_data['dest'] == 'SEA']
#proportion of originating airports
proportions = sea_flights['origin'].value_counts(normalize=True)
#from JFK and LGA while also preventing a keyerror if it can't find any
q_4 = proportions.reindex(['JFK', 'LGA'], fill_value=0.0)
print(f"q_4: The proportions of flights to SEA from JFK and LGA is {q_4}")
```

```python
datetime_series = pd.to_datetime(pd.DataFrame({
    'year': flights_data['year'],
    'month': flights_data['month'],
    'day': flights_data['day']
}))

flights_data["Date"] = datetime_series.dt.strftime("%Y/%m/%d")
avg_delays_dep = flights_data.groupby("Date")["dep_delay"].mean()
worst_date_dep = avg_delays_dep.idxmax()
max_delay_value = avg_delays_dep.max()
#q_5
print(f"q_5: The date with the largest average departure delay is: {worst_date}")
print(f"Average delay: {max_delay_value:.2f} minutes")
```
